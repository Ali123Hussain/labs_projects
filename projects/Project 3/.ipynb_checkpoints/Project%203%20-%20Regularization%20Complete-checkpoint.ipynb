# Regularization Models — Complete Code

This section contains a complete workflow for:

- Cleaning the Bahrain rental data
- Preparing numerical and categorical features
- Scaling the numerical features
- Building Ridge, Lasso, and Elastic Net models
- Comparing their performance
- Using cross-validation to select the best regularization values
- Training the final model
- Creating the submission file


```python
# 1. Import the required libraries

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

from sklearn.linear_model import (
    LinearRegression,
    Ridge,
    Lasso,
    ElasticNet,
    RidgeCV,
    LassoCV,
    ElasticNetCV
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Keep normal sklearn output to avoid issues with sparse encoded data
import sklearn
sklearn.set_config(transform_output="default")

pd.set_option("display.max_columns", None)
```

```python
# 2. Load the training data

DATA_PATH = "./data.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)
df.head()
```

```python
# 3. Check the data

print("Columns:")
print(df.columns.tolist())

print("\nMissing values:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nDuplicated rows:", df.duplicated().sum())

df.info()
```

```python
# 4. Remove rows where the target is missing

df = df.dropna(subset=["rent"]).copy()

# Remove duplicated rows
df = df.drop_duplicates().copy()

print("Shape after removing missing target values and duplicates:", df.shape)
```

```python
# 5. Clean the Beds column

# Convert special text values into reasonable numeric values
df["Beds"] = df["Beds"].replace({
    "studio": 1,
    "%+ Maid": 1,
    "studio+ Maid": 2,
    "%++ Maid": 1
})

# Extract the first number from values such as "7+"
df["Beds"] = pd.to_numeric(
    df["Beds"].astype(str).str.extract(r"(\d+)")[0],
    errors="coerce"
)

df["Beds"].value_counts(dropna=False).sort_index()
```

```python
# 6. Clean the Baths column

df["Baths"] = df["Baths"].replace({
    "none": 0
})

# Extract the first number from values such as "7+"
df["Baths"] = pd.to_numeric(
    df["Baths"].astype(str).str.extract(r"(\d+)")[0],
    errors="coerce"
)

df["Baths"].value_counts(dropna=False).sort_index()
```

```python
# 7. Clean the Size column

df["Size"] = (
    df["Size"]
    .astype(str)
    .str.lower()
    .str.replace(",", "", regex=False)
    .str.extract(r"(\d+\.?\d*)")[0]
)

df["Size"] = pd.to_numeric(df["Size"], errors="coerce")

df["Size"].describe()
```

```python
# 8. Clean the Amenities column

# Convert Amenities to numeric if possible.
# Missing or non-numeric values will become NaN and will later be imputed.
df["Amenities"] = pd.to_numeric(df["Amenities"], errors="coerce")

df["Amenities"].value_counts(dropna=False).head(10)
```

```python
# 9. Remove columns that should not be used as model features

columns_to_drop = [
    "Property_id",
    "URL",
    "Availability_date",
    "Title"
]

df = df.drop(
    columns=[column for column in columns_to_drop if column in df.columns]
)

print("Remaining columns:")
print(df.columns.tolist())
```

```python
# 10. Separate features and target

X = df.drop(columns=["rent"])
y = df["rent"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# 11. Split the data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
# 12. Identify numerical and categorical columns automatically

numeric_features = X.select_dtypes(include=["number"]).columns.tolist()
categorical_features = X.select_dtypes(exclude=["number"]).columns.tolist()

print("Numerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
# 13. Create the preprocessing pipelines

# Numerical columns:
# 1. Fill missing values with the median
# 2. Scale the values because regularization requires scaling
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

# Categorical columns:
# 1. Fill missing values with the most common value
# 2. Convert text categories into numeric dummy variables
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ],
    remainder="drop"
)

preprocessor
```

```python
# 14. Create a reusable evaluation function

def evaluate_regression_model(model_name, model, X_train, X_test, y_train, y_test):
    # Make predictions
    train_predictions = model.predict(X_train)
    test_predictions = model.predict(X_test)

    # Calculate metrics
    train_mae = mean_absolute_error(y_train, train_predictions)
    test_mae = mean_absolute_error(y_test, test_predictions)

    train_rmse = mean_squared_error(
        y_train,
        train_predictions
    ) ** 0.5

    test_rmse = mean_squared_error(
        y_test,
        test_predictions
    ) ** 0.5

    train_r2 = r2_score(y_train, train_predictions)
    test_r2 = r2_score(y_test, test_predictions)

    return {
        "Model": model_name,
        "Train MAE": train_mae,
        "Test MAE": test_mae,
        "Train RMSE": train_rmse,
        "Test RMSE": test_rmse,
        "Train R2": train_r2,
        "Test R2": test_r2
    }
```

## Baseline model

We first build ordinary Linear Regression. It gives us a baseline that we can compare with the regularized models.


```python
# 15. Build and train the baseline Linear Regression model

linear_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", LinearRegression())
    ]
)

linear_pipeline.fit(X_train, y_train)

linear_results = evaluate_regression_model(
    "Linear Regression",
    linear_pipeline,
    X_train,
    X_test,
    y_train,
    y_test
)

pd.DataFrame([linear_results])
```

## Ridge Regression — L2 Regularization

Ridge reduces the size of all coefficients but normally does not make them exactly zero.


```python
# 16. Build and train a basic Ridge model

ridge_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", Ridge(alpha=1.0))
    ]
)

ridge_pipeline.fit(X_train, y_train)

ridge_results = evaluate_regression_model(
    "Ridge",
    ridge_pipeline,
    X_train,
    X_test,
    y_train,
    y_test
)

pd.DataFrame([ridge_results])
```

## Lasso Regression — L1 Regularization

Lasso can reduce some coefficients to exactly zero, so it can also perform feature selection.


```python
# 17. Build and train a basic Lasso model

lasso_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "regressor",
            Lasso(
                alpha=1.0,
                max_iter=20000
            )
        )
    ]
)

lasso_pipeline.fit(X_train, y_train)

lasso_results = evaluate_regression_model(
    "Lasso",
    lasso_pipeline,
    X_train,
    X_test,
    y_train,
    y_test
)

pd.DataFrame([lasso_results])
```

## Elastic Net

Elastic Net combines Ridge and Lasso penalties.

- `alpha` controls the overall regularization strength.
- `l1_ratio` controls the balance between Lasso and Ridge.
- `l1_ratio = 1` behaves like Lasso.
- `l1_ratio = 0` behaves like Ridge.


```python
# 18. Build and train a basic Elastic Net model

elastic_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "regressor",
            ElasticNet(
                alpha=1.0,
                l1_ratio=0.5,
                max_iter=20000
            )
        )
    ]
)

elastic_pipeline.fit(X_train, y_train)

elastic_results = evaluate_regression_model(
    "Elastic Net",
    elastic_pipeline,
    X_train,
    X_test,
    y_train,
    y_test
)

pd.DataFrame([elastic_results])
```

```python
# 19. Compare the four initial models

initial_results = pd.DataFrame([
    linear_results,
    ridge_results,
    lasso_results,
    elastic_results
])

initial_results.sort_values(
    by="Test RMSE"
).reset_index(drop=True)
```

# Cross-validation

Instead of manually choosing one value for `alpha`, cross-validation tests several values and selects the best one.


```python
# 20. Values that the models will test

alpha_values = np.logspace(-4, 4, 100)

print(alpha_values[:5])
print(alpha_values[-5:])
```

```python
# 21. RidgeCV

ridge_cv_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "regressor",
            RidgeCV(
                alphas=alpha_values,
                cv=5,
                scoring="neg_mean_squared_error"
            )
        )
    ]
)

ridge_cv_pipeline.fit(X_train, y_train)

best_ridge_alpha = ridge_cv_pipeline.named_steps["regressor"].alpha_

print("Best Ridge alpha:", best_ridge_alpha)
```

```python
# 22. Evaluate RidgeCV

ridge_cv_results = evaluate_regression_model(
    "RidgeCV",
    ridge_cv_pipeline,
    X_train,
    X_test,
    y_train,
    y_test
)

pd.DataFrame([ridge_cv_results])
```

```python
# 23. LassoCV

lasso_cv_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "regressor",
            LassoCV(
                alphas=alpha_values,
                cv=5,
                max_iter=50000,
                random_state=42
            )
        )
    ]
)

lasso_cv_pipeline.fit(X_train, y_train)

best_lasso_alpha = lasso_cv_pipeline.named_steps["regressor"].alpha_

print("Best Lasso alpha:", best_lasso_alpha)
```

```python
# 24. Evaluate LassoCV

lasso_cv_results = evaluate_regression_model(
    "LassoCV",
    lasso_cv_pipeline,
    X_train,
    X_test,
    y_train,
    y_test
)

pd.DataFrame([lasso_cv_results])
```

```python
# 25. ElasticNetCV

elastic_cv_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "regressor",
            ElasticNetCV(
                alphas=alpha_values,
                l1_ratio=[0.1, 0.3, 0.5, 0.7, 0.9, 0.95, 0.99],
                cv=5,
                max_iter=50000,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

elastic_cv_pipeline.fit(X_train, y_train)

best_elastic_alpha = elastic_cv_pipeline.named_steps["regressor"].alpha_
best_l1_ratio = elastic_cv_pipeline.named_steps["regressor"].l1_ratio_

print("Best Elastic Net alpha:", best_elastic_alpha)
print("Best Elastic Net l1_ratio:", best_l1_ratio)
```

```python
# 26. Evaluate ElasticNetCV

elastic_cv_results = evaluate_regression_model(
    "ElasticNetCV",
    elastic_cv_pipeline,
    X_train,
    X_test,
    y_train,
    y_test
)

pd.DataFrame([elastic_cv_results])
```

```python
# 27. Compare all models

all_results = pd.DataFrame([
    linear_results,
    ridge_results,
    lasso_results,
    elastic_results,
    ridge_cv_results,
    lasso_cv_results,
    elastic_cv_results
])

all_results = all_results.sort_values(
    by="Test RMSE"
).reset_index(drop=True)

all_results
```

```python
# 28. Display the best model according to Test RMSE

best_model_name = all_results.loc[0, "Model"]
best_test_rmse = all_results.loc[0, "Test RMSE"]
best_test_mae = all_results.loc[0, "Test MAE"]
best_test_r2 = all_results.loc[0, "Test R2"]

print("Best model:", best_model_name)
print("Best Test RMSE:", best_test_rmse)
print("Best Test MAE:", best_test_mae)
print("Best Test R2:", best_test_r2)
```

# Inspect the coefficients

This allows us to see which transformed features have the largest positive or negative effects.


```python
# 29. Choose one regularized model for coefficient inspection

selected_regularized_model = elastic_cv_pipeline

feature_names = (
    selected_regularized_model
    .named_steps["preprocessor"]
    .get_feature_names_out()
)

coefficients = (
    selected_regularized_model
    .named_steps["regressor"]
    .coef_
)

coefficient_df = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients,
    "Absolute Coefficient": np.abs(coefficients)
})

coefficient_df = coefficient_df.sort_values(
    by="Absolute Coefficient",
    ascending=False
)

coefficient_df.head(20)
```

```python
# 30. Count how many coefficients Elastic Net reduced to zero

zero_coefficients = (coefficients == 0).sum()
non_zero_coefficients = (coefficients != 0).sum()

print("Zero coefficients:", zero_coefficients)
print("Non-zero coefficients:", non_zero_coefficients)
```
