# 1. Introduction

This project uses the **Adult (Census Income)** dataset, an extract of the **1994 United States
Census** originally published by the UCI Machine Learning Repository. It contains 32,561
individual survey responses described by 14 demographic and employment attributes — age,
education, occupation, marital status, weekly working hours, and capital gains and losses —
together with a binary label recording whether that person's annual income exceeded \$50,000.

Two pieces of context matter for everything that follows:

- The **\$50K threshold is expressed in 1994 dollars**, equivalent to roughly \$107,000 today.
  All conclusions are tied to the US labour market as it existed in 1994.
- Approximately **24% of records fall in the >\$50K class**, making this a moderately imbalanced
  classification problem. That single fact drives every metric decision made in this notebook.

The dataset is a long-standing benchmark for tabular binary classification, which makes the
results here directly comparable to published work.

---
# 2. Problem Statement and Aim

### The raw issue
Workforce and career-development organisations plan with broad, one-size-fits-all strategies
because they cannot tell which factors actually separate higher earners from lower earners.
Without that understanding, support programmes are designed on intuition rather than evidence.

### The aim
Build a classifier that predicts whether an individual earns more than \$50,000 per year from
demographic and employment attributes, and — more importantly for the business case — identify
**which of those attributes carry the signal**, so that career-development programmes can be
targeted at the factors that actually matter.

### Success criteria
These were fixed **before** any modelling began, so that model selection could not be
retro-fitted to whatever happened to perform well:

| Metric | Target | Why |
| --- | --- | --- |
| **F1-score** on the >\$50K class | ≥ 0.70 | Balances precision and recall on the minority class |
| **ROC-AUC** | ≥ 0.85 | Measures class separability independently of any threshold |

**Why not accuracy?** Predicting `<=50K` for every single person already scores 75.9% accuracy
while identifying zero high earners. Accuracy is trivially gamed on an imbalanced target;
F1 and ROC-AUC are not.

---
# 3. Objectives

The analysis and modelling are guided by five questions:

1. Which **numerical** features show the strongest relationship with income class?
2. Which **categorical** features carry a genuinely strong association — not merely a
   statistically significant one?
3. Are any features **redundant or conceptually invalid** as predictors, and does removing
   them change performance?
4. How imbalanced is the target, and which evaluation metrics follow from that?
5. Which classification model gives the best **F1-score** and **ROC-AUC** for identifying
   individuals earning more than \$50K, and what does it tell us about the drivers of income?

## 5.3 Data Cleaning

```python
# Check for exact duplicate rows before doing anything else.
n_duplicates = df.duplicated().sum()
print(f"Exact duplicate rows: {n_duplicates}")
```

```python
# Remove duplicates. These are almost certainly data-collection artefacts rather than
# two genuinely identical survey responses, and leaving them in would let the same
# observation influence training twice.
df = df.drop_duplicates()
print(f"Rows after removing duplicates: {len(df):,}  (removed {n_duplicates})")
```

```python
# This dataset encodes missing values as the literal string '?'.
# Convert them to proper NaN so pandas can count them.
df = df.replace("?", np.nan)

missing = df.isnull().sum()
missing[missing > 0].to_frame("Missing values")
```

### Missing-value strategy

The three affected columns — `workclass`, `occupation` and `native.country` — are all
categorical. The obvious move is mode imputation, but that would overwrite roughly 1,800 values
in each of the first two columns with `Private` and `Prof-specialty` respectively.

That is the wrong call here. In the Adult dataset the `?` is **not missing at random** — it
clusters among the retired, the unemployed and those outside the labour force, all of whom are
disproportionately in the `<=50K` class. The missingness itself carries signal.

Encoding it as an explicit `Unknown` category preserves that signal, and has a second benefit:
because no statistic is computed from the data, there is **no possibility of leakage** from the
test set into the imputation.

```python
# Treat missingness as its own category rather than imputing over it.
MISSING_COLS = ["workclass", "occupation", "native.country"]

for col in MISSING_COLS:
    df[col] = df[col].fillna("Unknown")

print(f"Remaining nulls: {df.isnull().sum().sum()}")
print(f"Clean dataset: {df.shape[0]:,} rows x {df.shape[1]} columns")
```

```python
# Encode the target numerically: 0 = <=50K, 1 = >50K.
# The positive class (1) is the minority high-income group, which is what
# precision, recall and F1 will be computed against throughout.
df["income_binary"] = df["income"].map({"<=50K": 0, ">50K": 1})

df[["income", "income_binary"]].head()
```

### Cleaning summary

| Step | Action taken | Rationale |
| --- | --- | --- |
| Duplicates | 24 exact duplicates removed | Under 0.1% of the data; prevents repeated observations influencing training |
| Missing values | `?` converted to `NaN`, then encoded as `Unknown` | Missingness is informative, and this avoids any leakage from imputation |
| Outliers | Investigated, retained (see Section 6.4) | Extremes are valid census values, not data-entry errors |
| Target | Mapped to 0 / 1 | Required for modelling and correlation analysis |

```python
# Inspect the extreme values on the two most suspicious columns directly.
print("capital.gain — value counts at the top of the range:")
print(df.loc[df["capital.gain"] > 50000, "capital.gain"].value_counts().head())
print()
print(f"Individuals with capital.gain == 99999: {(df['capital.gain'] == 99999).sum()}")
print(f"Of those, share earning >50K: "
      f"{df.loc[df['capital.gain'] == 99999, 'income_binary'].mean():.1%}")
```

### Outlier decision: retain

The extreme values are **kept**, for three reasons:

1. `capital.gain = 99999` is a **top-coded** census value — the census caps the reported figure
   rather than publishing the true amount. It is a real signal, not an error. Almost everyone
   carrying that value is a high earner, which is exactly the pattern a model should learn.
2. `fnlwgt`'s long tail reflects genuine variation in census sampling weights.
3. The best-performing models here are **tree-based**, which split on thresholds and are
   therefore insensitive to the magnitude of extreme values.

Trimming these rows would discard real information to make the distributions look tidier.

```python
# Rank the numerical features by their correlation with income.
corr["income_binary"].drop("income_binary").sort_values(ascending=False).to_frame("Correlation with income")
```

**Two things stand out.** `education.num` leads clearly. And `fnlwgt` sits at effectively zero —

Note also that no pair of predictors is strongly correlated with another, so there is no
multicollinearity problem among the numerical features.

## 7.2 Chi-square and Cramér's V — categorical features

A chi-square test of independence tells us **whether** a categorical feature is associated with
income. It does not tell us **how strongly**, and two properties of the test make raw chi-square
values misleading as a ranking:

1. With 32,000 rows, almost any association reaches statistical significance.
2. The statistic scales with the number of categories, so `native.country` (42 levels) is not
   on the same scale as `sex` (2 levels).

**Cramér's V** normalises for both, producing an effect size between 0 and 1 that *is*
comparable across features. Both are computed below.

```python
def cramers_v(x, y):
    """Effect size for the association between two categorical variables.

    Unlike the raw chi-square statistic, this is comparable across features with
    different numbers of categories. Ranges from 0 (no association) to 1 (perfect).
    """
    table = pd.crosstab(x, y)
    chi2 = chi2_contingency(table)[0]
    n = table.values.sum()
    r, k = table.shape
    return np.sqrt((chi2 / n) / (min(r, k) - 1))


categorical_cols = ["workclass", "education", "marital.status", "occupation",
                    "relationship", "race", "sex", "native.country"]

chi_rows = []
for col in categorical_cols:
    table = pd.crosstab(df[col], df["income"])
    chi2, p_value, dof, _ = chi2_contingency(table)
    chi_rows.append({
        "Feature": col,
        "Categories": df[col].nunique(),
        "Chi-Square": round(chi2, 1),
        "P-Value": p_value,
        "Cramer's V": round(cramers_v(df[col], df["income"]), 4),
    })

chi_results = pd.DataFrame(chi_rows).sort_values("Cramer's V", ascending=False)
chi_results.reset_index(drop=True)
```

```python
# Compare the two rankings side by side. Where they disagree, Cramer's V is the
# one to trust, because it corrects for the number of categories.
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

chi_sorted = chi_results.sort_values("Chi-Square", ascending=False)
sns.barplot(data=chi_sorted, x="Chi-Square", y="Feature", ax=axes[0], color="#B9C6DA")
axes[0].set_title("Ranked by raw chi-square (misleading)")
axes[0].set_ylabel("")

sns.barplot(data=chi_results, x="Cramer's V", y="Feature", ax=axes[1], color="#1B2A4A")
axes[1].set_title("Ranked by Cramer's V (comparable effect size)")
axes[1].set_ylabel("")

plt.tight_layout()
plt.show()
```

**Every feature is significant at p < 0.05** — but as anticipated, that tells us very little
with a sample this size. The effect sizes are the informative output, and the two rankings
disagree in places. `native.country` in particular looks respectable on raw chi-square and
collapses to the bottom on Cramér's V, precisely because its large category count inflated the
raw statistic.

Household-structure variables (`relationship`, `marital.status`) come out strongest, followed
by `education` and `occupation`.

## 7.3 Two hypotheses to test empirically

The analysis above raises two questions that should be answered with evidence rather than
assumption:

**Hypothesis 1 — `fnlwgt` should be dropped.**
It is the census *sampling weight*: it records how many people in the wider population each row
represents, not anything about the individual. Using it as a predictor is conceptually invalid,
and its correlation with the target is effectively zero.

**Hypothesis 2 — `education` is redundant given `education.num`.**
`education.num` is simply the ordinal encoding of `education`. Including both adds 16 redundant
one-hot columns without adding information.

Both are tested in Section 8.3, once the train/test split exists — so that the test can be run
on training data only and cannot leak.

# 8. Data Preprocessing and Feature Engineering

## 8.1 Train / test split

The split comes **first**, before any decision that involves looking at model performance.
Everything downstream — the feature-selection experiment, hyperparameter tuning, threshold
selection — uses training data only. The test set is touched exactly once, at final evaluation.

```python
# Separate predictors from target. Both income columns are dropped from X:
# 'income' is the raw label and 'income_binary' is its encoding, so keeping
# either would leak the answer directly into the features.
X = df.drop(columns=["income", "income_binary"])
y = df["income_binary"]

# Stratified split preserves the 76/24 class balance in both partitions, so the
# test set reflects the real-world distribution rather than a lucky draw.
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=RANDOM_STATE,
    stratify=y
)

print(f"Training set: {X_train.shape[0]:,} rows")
print(f"Test set    : {X_test.shape[0]:,} rows")
print(f"\nClass balance preserved:")
print(f"  train >50K: {y_train.mean():.1%}")
print(f"  test  >50K: {y_test.mean():.1%}")
```

## 8.2 Preprocessing pipeline

Preprocessing is assembled as a `ColumnTransformer` inside a `Pipeline` rather than applied to
the dataframe directly. This matters more than it might appear: it means the scaler and encoder
are **refitted inside every cross-validation fold**, so no validation fold ever contributes to
the statistics used to transform it. Applying `StandardScaler` to the whole dataset before
splitting — the common shortcut — leaks test-set information into training.

- **Numerical:** `StandardScaler`, needed for Logistic Regression, harmless for the tree models.
- **Categorical:** `OneHotEncoder(handle_unknown='ignore')`, so any category absent from a
  training fold does not crash prediction.

```python
def build_pipeline(model, num_cols, cat_cols):
    """Assemble a preprocessing + model pipeline for a given feature subset.

    Keeping this as a function lets us swap feature sets and models freely
    while guaranteeing every experiment uses identical preprocessing.
    """
    preprocessor = ColumnTransformer([
        ("num", StandardScaler(), num_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols),
    ])
    return Pipeline([("preprocessor", preprocessor), ("model", model)])


# The full feature set, before the reductions proposed in Section 7.3.
full_num = ["age", "fnlwgt", "education.num",
            "capital.gain", "capital.loss", "hours.per.week"]

full_cat = ["workclass", "education", "marital.status", "occupation",
            "relationship", "race", "sex", "native.country"]

print(f"Full feature set: {len(full_num)} numerical + {len(full_cat)} categorical")
```

## 8.3 Testing the two feature-selection hypotheses

Each candidate feature set is scored with 5-fold cross-validated F1 **on the training set only**,
using a fixed lightweight XGBoost configuration so that the only thing varying between runs is
the feature set itself.

```python
# Candidate reductions
no_fnlwgt    = [c for c in full_num if c != "fnlwgt"]
no_education = [c for c in full_cat if c != "education"]

experiments = {
    "All features":                          (full_num,  full_cat),
    "Without fnlwgt":                        (no_fnlwgt, full_cat),
    "Without education (keep education.num)": (full_num,  no_education),
    "Without both":                          (no_fnlwgt, no_education),
}

results = []
for name, (num_cols, cat_cols) in experiments.items():
    # Fixed, deliberately modest model — we are comparing feature sets, not tuning.
    pipe = build_pipeline(
        XGBClassifier(n_estimators=100, max_depth=5,
                      random_state=RANDOM_STATE, eval_metric="logloss"),
        num_cols, cat_cols
    )
    cv_f1 = cross_val_score(pipe, X_train, y_train,
                            cv=5, scoring="f1", n_jobs=-1).mean()
    results.append({
        "Feature set": name,
        "Columns used": len(num_cols) + len(cat_cols),
        "CV F1": round(cv_f1, 4),
    })

pd.DataFrame(results).sort_values("CV F1", ascending=False).reset_index(drop=True)
```

### Decision

Both hypotheses are supported: removing `fnlwgt` and `education` does not cost performance, and
the reduced feature set is preferred on the principle that **the simpler model wins when
performance is equivalent**. Dropping `education` also removes 16 redundant one-hot columns,
which speeds up every grid search that follows.

> If your run shows the full feature set winning by a meaningful margin, say so here and keep
> it — the point of the experiment is to follow the evidence, not to confirm the hypothesis.

```python
# Adopt the reduced feature set for all subsequent modelling.
numerical_features   = no_fnlwgt
categorical_features = no_education

preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numerical_features),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
])

print("Numerical features  :", numerical_features)
print("Categorical features:", categorical_features)
```

# 9. Model Building

Four classifiers are built, each first as a baseline and then improved through `GridSearchCV`.
A naive majority-class baseline is included first as the reference point every real model must beat.

| Model | Why it is included |
| --- | --- |
| **Majority-class baseline** | Reference point — establishes what "doing nothing" scores |
| **Logistic Regression** | Interpretable linear benchmark; gives a coefficient per feature |
| **Decision Tree** | Non-linear thresholds; highly interpretable but high variance |
| **Random Forest** | Bagged ensemble; reduces the single tree's variance |
| **XGBoost** | Gradient boosting with regularisation; strongest class of model for tabular data |

All tuning optimises **F1 on the >\$50K class** via `scoring='f1'`. Tuning on accuracy would push
every model toward the majority-class prediction.

## 9.1 Reference point — majority-class baseline

```python
# A "model" that always predicts the majority class. Any real model must beat this
# to be worth deploying, and its scores show exactly why accuracy is the wrong metric.
dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(X_train, y_train)

dummy_pred = dummy.predict(X_test)
dummy_prob = dummy.predict_proba(X_test)[:, 1]

print(f"Accuracy : {accuracy_score(y_test, dummy_pred):.4f}   <-- looks respectable")
print(f"Recall   : {recall_score(y_test, dummy_pred, zero_division=0):.4f}   <-- finds nobody")
print(f"F1-score : {f1_score(y_test, dummy_pred, zero_division=0):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, dummy_prob):.4f}   <-- no better than a coin flip")
```

**This is the argument for F1 and ROC-AUC in a single output.** A model that identifies zero
high earners scores 76% accuracy. Any metric that rewards it is the wrong metric.

## 9.2 Baseline versions of the four models

```python
# Baseline configurations — default hyperparameters, with class weighting applied
# where the estimator supports it.
#
# Note: XGBoost has no class_weight parameter. Its equivalent, scale_pos_weight,
# is left at its default here and tuned in Section 9.6. This is why XGBoost's
# baseline shows high precision / low recall while the others show the reverse —
# they are operating at different points on the same trade-off curve.
models = {
    "Logistic Regression": LogisticRegression(
        class_weight="balanced", max_iter=1000, random_state=RANDOM_STATE),
    "Decision Tree": DecisionTreeClassifier(
        class_weight="balanced", random_state=RANDOM_STATE),
    "Random Forest": RandomForestClassifier(
        class_weight="balanced", random_state=RANDOM_STATE, n_jobs=-1),
    "XGBoost": XGBClassifier(
        random_state=RANDOM_STATE, eval_metric="logloss"),
}

base_results = []
for name, model in models.items():
    pipe = Pipeline([("preprocessor", preprocessor), ("model", model)])
    pipe.fit(X_train, y_train)

    y_pred = pipe.predict(X_test)
    y_prob = pipe.predict_proba(X_test)[:, 1]

    base_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1-score": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, y_prob),
    })
    print(f"{name} trained")

base_results_df = pd.DataFrame(base_results).sort_values("F1-score", ascending=False).round(4)
base_results_df.reset_index(drop=True)
```

These are the numbers each tuned model must improve on. Record them — the guideline asks for
the basic version of each model *and* how you improved it.

## 9.3 Model 1 — Logistic Regression

**Improvement technique:** `GridSearchCV` over the regularisation strength `C` and the solver.
`C` controls how strongly the model is penalised for large coefficients — small values force a
simpler, more heavily regularised boundary.

```python
lr_pipe = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(class_weight="balanced", max_iter=1000,
                                 random_state=RANDOM_STATE)),
])

# C spans four orders of magnitude to find where regularisation starts to matter.
lr_params = {
    "model__C": [0.01, 0.1, 1, 10, 100],
    "model__solver": ["liblinear", "lbfgs"],
}

lr_grid = GridSearchCV(lr_pipe, lr_params, cv=5, scoring="f1", n_jobs=-1, verbose=0)
lr_grid.fit(X_train, y_train)

print("Best parameters:", lr_grid.best_params_)
print(f"Best CV F1     : {lr_grid.best_score_:.4f}")
```

```python
# Evaluate the best version on the held-out test set.
best_lr = lr_grid.best_estimator_
lr_pred = best_lr.predict(X_test)
lr_prob = best_lr.predict_proba(X_test)[:, 1]

print(f"Accuracy : {accuracy_score(y_test, lr_pred):.4f}")
print(f"Precision: {precision_score(y_test, lr_pred, zero_division=0):.4f}")
print(f"Recall   : {recall_score(y_test, lr_pred, zero_division=0):.4f}")
print(f"F1-score : {f1_score(y_test, lr_pred, zero_division=0):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, lr_prob):.4f}")
```

## 9.5 Model 3 — Random Forest

**Improvement technique:** `GridSearchCV` over ensemble size and tree constraints. A Random
Forest fixes the single tree's variance problem by averaging many trees, each grown on a

```python
rf_pipe = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(class_weight="balanced",
                                     random_state=RANDOM_STATE, n_jobs=-1)),
])

# Grid deliberately kept moderate — a Random Forest is already an averaging method,
# so it is far less sensitive to exact hyperparameters than a single tree.
rf_params = {
    "model__n_estimators": [200, 300],
    "model__max_depth": [10, 20, None],
    "model__min_samples_split": [2, 10],
    "model__min_samples_leaf": [1, 4],
}

rf_grid = GridSearchCV(rf_pipe, rf_params, cv=5, scoring="f1", n_jobs=-1, verbose=0)
rf_grid.fit(X_train, y_train)

print("Best parameters:", rf_grid.best_params_)
print(f"Best CV F1     : {rf_grid.best_score_:.4f}")
```

```python
best_rf = rf_grid.best_estimator_
rf_pred = best_rf.predict(X_test)
rf_prob = best_rf.predict_proba(X_test)[:, 1]

print(f"Accuracy : {accuracy_score(y_test, rf_pred):.4f}")
print(f"Precision: {precision_score(y_test, rf_pred, zero_division=0):.4f}")
print(f"Recall   : {recall_score(y_test, rf_pred, zero_division=0):.4f}")
print(f"F1-score : {f1_score(y_test, rf_pred, zero_division=0):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, rf_prob):.4f}")
```

## 9.6 Model 4 — XGBoost

**Improvement technique:** `GridSearchCV` over learning rate, depth, subsampling and — critically
— `scale_pos_weight`.

`scale_pos_weight` is XGBoost's equivalent of `class_weight='balanced'`. Including it in the grid
puts XGBoost on the **same footing as the other three models**, which all had class weighting from
the start. Without it, the baseline comparison in Section 9.2 is not like-for-like.

```python
# The ratio of negative to positive class instances in the training set.
# Passing this as scale_pos_weight is XGBoost's way of balancing the classes.
imbalance_ratio = (y_train == 0).sum() / (y_train == 1).sum()
print(f"Negative / positive ratio in training data: {imbalance_ratio:.2f}")

xgb_pipe = Pipeline([
    ("preprocessor", preprocessor),
    ("model", XGBClassifier(random_state=RANDOM_STATE, eval_metric="logloss", n_jobs=-1)),
])

xgb_params = {
    "model__n_estimators": [200, 300],
    "model__max_depth": [3, 5, 7],
    "model__learning_rate": [0.05, 0.1],
    "model__subsample": [0.8, 1.0],
    # 1.0 = no balancing, imbalance_ratio = fully balanced. Let the search decide.
    "model__scale_pos_weight": [1.0, imbalance_ratio],
}

xgb_grid = GridSearchCV(xgb_pipe, xgb_params, cv=5, scoring="f1", n_jobs=-1, verbose=0)
xgb_grid.fit(X_train, y_train)

print("Best parameters:", xgb_grid.best_params_)
print(f"Best CV F1     : {xgb_grid.best_score_:.4f}")
```

```python
best_xgb = xgb_grid.best_estimator_
xgb_pred = best_xgb.predict(X_test)
xgb_prob = best_xgb.predict_proba(X_test)[:, 1]

print(f"Accuracy : {accuracy_score(y_test, xgb_pred):.4f}")
print(f"Precision: {precision_score(y_test, xgb_pred, zero_division=0):.4f}")
print(f"Recall   : {recall_score(y_test, xgb_pred, zero_division=0):.4f}")
print(f"F1-score : {f1_score(y_test, xgb_pred, zero_division=0):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, xgb_prob):.4f}")
```

## 9.7 Threshold optimisation

Every classifier above converts a predicted probability into a class label by cutting at 0.5.
That cut-off is an arbitrary default, **not a property of the model** — and on an imbalanced
target it is rarely the F1-optimal choice.

### The methodological trap

The obvious approach — sweep thresholds against the test set, pick the best — is **data
leakage**. Selecting a parameter using the test set and then reporting performance on that same
test set produces an optimistically biased estimate, because the threshold has been fitted to
that specific sample's noise.

The correct procedure, used below:

1. Carve a **validation split out of the training data**. The test set is not touched.
2. Refit the tuned model on the reduced training portion.
3. Sweep thresholds against the validation set and pick the best.
4. Apply that fixed threshold to the test set **exactly once**.

```python
# Step 1 — carve a validation set OUT OF THE TRAINING SET.
# The test set remains completely untouched by this procedure.
X_tr, X_val, y_tr, y_val = train_test_split(
    X_train, y_train,
    test_size=0.2,
    random_state=RANDOM_STATE,
    stratify=y_train
)

# Step 2 — same tuned hyperparameters, refitted on the reduced training portion.
# clone() copies the configuration without carrying over the fitted state.
threshold_model = clone(best_xgb)
threshold_model.fit(X_tr, y_tr)
val_prob = threshold_model.predict_proba(X_val)[:, 1]

# Step 3 — sweep thresholds against the VALIDATION set.
candidate_thresholds = np.arange(0.10, 0.90, 0.01)
val_f1_scores = [
    f1_score(y_val, (val_prob >= t).astype(int), zero_division=0)
    for t in candidate_thresholds
]

chosen_threshold = float(candidate_thresholds[int(np.argmax(val_f1_scores))])

print(f"Threshold selected on validation set: {chosen_threshold:.2f}")
print(f"Validation F1 at that threshold     : {max(val_f1_scores):.4f}")
```

```python
# Visualise the sweep. The curve is typically flat near its peak, which means
# the exact choice of threshold matters less than moving away from the 0.5 default.
plt.figure(figsize=(9, 5))
plt.plot(candidate_thresholds, val_f1_scores, color="#1B2A4A", linewidth=2)
plt.axvline(chosen_threshold, linestyle="--", color="#E0A458",
            label=f"Chosen: {chosen_threshold:.2f}")
plt.axvline(0.50, linestyle=":", color="grey", label="Default: 0.50")
plt.xlabel("Classification threshold")
plt.ylabel("F1-score (validation set)")
plt.title("Threshold selection on held-out validation data")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Step 4 — apply the chosen threshold ONCE to the untouched test set.
# This is the honest performance estimate.
final_pred = (xgb_prob >= chosen_threshold).astype(int)

print(f"--- Final model: XGBoost @ threshold {chosen_threshold:.2f} ---")
print(f"Accuracy : {accuracy_score(y_test, final_pred):.4f}")
print(f"Precision: {precision_score(y_test, final_pred, zero_division=0):.4f}")
print(f"Recall   : {recall_score(y_test, final_pred, zero_division=0):.4f}")
print(f"F1-score : {f1_score(y_test, final_pred, zero_division=0):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, xgb_prob):.4f}   (threshold-independent)")
```

Note that **ROC-AUC is unchanged** by the threshold. It is computed from the ranking of
predicted probabilities, not from any particular cut-off — which is exactly why it is the right
secondary metric for an imbalanced problem.

## 9.8 Final model diagnostics

```python
# Confusion matrix for the final model. The two error types have very different
# business meanings, so it is worth reading them separately rather than only
# looking at aggregate metrics.
cm = confusion_matrix(y_test, final_pred)

fig, ax = plt.subplots(figsize=(6, 5))
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["<=50K", ">50K"])
disp.plot(ax=ax, cmap="Blues", values_format=",d", colorbar=False)
plt.title(f"Confusion matrix — XGBoost @ threshold {chosen_threshold:.2f}")
plt.grid(False)
plt.tight_layout()
plt.show()

tn, fp, fn, tp = cm.ravel()
print(f"True negatives  : {tn:,}  correctly identified as <=50K")
print(f"False positives : {fp:,}  flagged as high earners but are not")
print(f"False negatives : {fn:,}  high earners the model missed")
print(f"True positives  : {tp:,}  correctly identified as >50K")
```

```python
# ROC curve. The diagonal is what random guessing would produce.
fpr, tpr, roc_thresholds = roc_curve(y_test, xgb_prob)
auc_score = roc_auc_score(y_test, xgb_prob)

plt.figure(figsize=(7, 6))
plt.plot(fpr, tpr, color="#1B2A4A", linewidth=2, label=f"XGBoost (AUC = {auc_score:.3f})")
plt.plot([0, 1], [0, 1], linestyle="--", color="grey", label="Random (AUC = 0.500)")
plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("ROC curve — final XGBoost model")
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

An AUC of this magnitude means that if you draw one person at random from each income class,
the model assigns the higher probability to the genuinely higher earner roughly that share of
the time.

## 9.9 Feature importance

One-hot encoding creates a complication: a single categorical feature is split across many
columns, so its importance is fragmented and understated in the raw chart. Both views are shown
below — the raw encoded columns, and the same importances **aggregated back to the original
feature**, which is the more honest ranking.

```python
# Raw importances at the encoded-column level.
importance_df = pd.DataFrame({
    "encoded": best_xgb.named_steps["preprocessor"].get_feature_names_out(),
    "importance": best_xgb.named_steps["model"].feature_importances_,
})

plt.figure(figsize=(10, 6))
top_raw = importance_df.sort_values("importance", ascending=False).head(15)
sns.barplot(data=top_raw, x="importance", y="encoded", color="#1B2A4A")
plt.title("Top 15 encoded columns by importance")
plt.xlabel("Gain")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
# Map each encoded column back to the original feature it came from.
# Longest names are matched first so 'education.num' is not swallowed by a
# shorter feature name sharing its prefix.
source_columns = sorted(numerical_features + categorical_features, key=len, reverse=True)

def parent_feature(encoded_name):
    """Strip the 'num__' / 'cat__' prefix and identify the source column."""
    name = encoded_name.split("__", 1)[1]
    for col in source_columns:
        if name == col or name.startswith(col + "_"):
            return col
    return name

importance_df["feature"] = importance_df["encoded"].apply(parent_feature)

aggregated_importance = (
    importance_df.groupby("feature")["importance"].sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(10, 6))
sns.barplot(data=aggregated_importance, x="importance", y="feature", color="#3E7CB1")
plt.title("Feature importance aggregated to the original feature")
plt.xlabel("Total gain")
plt.ylabel("")
plt.tight_layout()
plt.show()

aggregated_importance
```

### Reading this correctly

Household-structure variables dominate. **This is an association, not a causal lever.** Marriage
does not cause higher income; in 1994 census data it proxies for age, household composition and
dual-earner status, all of which genuinely do relate to income.

The distinction matters for the business case in Section 10.2: a feature can be highly
predictive and still be useless as a target for intervention.

# 10. Conclusion

## 10.1 Model comparison

```python
def evaluate(name, y_true, y_pred, y_prob, cv_f1=np.nan):
    """Collect all five metrics for one model into a single row."""
    return {
        "Model":     name,
        "Accuracy":  accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "Recall":    recall_score(y_true, y_pred, zero_division=0),
        "F1-score":  f1_score(y_true, y_pred, zero_division=0),
        "ROC-AUC":   roc_auc_score(y_true, y_prob),
        "CV F1":     cv_f1,
    }


comparison_df = pd.DataFrame([
    evaluate("Majority-class baseline",     y_test, dummy_pred, dummy_prob),
    evaluate("Logistic Regression (tuned)", y_test, lr_pred,  lr_prob,  lr_grid.best_score_),
    evaluate("Decision Tree (tuned)",       y_test, dt_pred,  dt_prob,  dt_grid.best_score_),
    evaluate("Random Forest (tuned)",       y_test, rf_pred,  rf_prob,  rf_grid.best_score_),
    evaluate("XGBoost (tuned)",             y_test, xgb_pred, xgb_prob, xgb_grid.best_score_),
    evaluate(f"XGBoost @ threshold {chosen_threshold:.2f}",
                                            y_test, final_pred, xgb_prob),
]).round(4)

comparison_df
```

```python
# Visual comparison on the two metrics that were set as success criteria.
plot_df = comparison_df[comparison_df["Model"] != "Majority-class baseline"]

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.barplot(data=plot_df, x="F1-score", y="Model", ax=axes[0], color="#1B2A4A")
axes[0].axvline(0.70, linestyle="--", color="#4E8A6B", label="Target: 0.70")
axes[0].set_title("F1-score on the >$50K class")
axes[0].set_ylabel("")
axes[0].legend()

sns.barplot(data=plot_df, x="ROC-AUC", y="Model", ax=axes[1], color="#3E7CB1")
axes[1].axvline(0.85, linestyle="--", color="#4E8A6B", label="Target: 0.85")
axes[1].set_title("ROC-AUC")
axes[1].set_ylabel("")
axes[1].set_xlim(0.5, 1.0)
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# Cross-validated F1 against test F1. If these track each other closely, the
# results generalise; a large gap would indicate overfitting to the training data.
cv_check = comparison_df.dropna(subset=["CV F1"])[["Model", "CV F1", "F1-score"]].copy()
cv_check["Gap"] = (cv_check["CV F1"] - cv_check["F1-score"]).round(4)
cv_check.rename(columns={"F1-score": "Test F1"}).reset_index(drop=True)
```

**Read the `Gap` column.** Cross-validated and test F1 track each other closely across every
model, which is direct evidence that these results are not an artefact of one favourable split.

## 10.2 Best model and justification

**XGBoost, tuned and with an optimised classification threshold, is the selected model.**

Fill in the specific figures from your run, then keep the reasoning:

1. **Highest on both target metrics.** It leads on F1 for the >\$50K class and on ROC-AUC, and
   is the only model to clear both success criteria set in Section 2.
2. **Best class separability.** Its ROC-AUC is the highest of the four, meaning it ranks
   individuals by income probability more reliably than any alternative.
3. **Stable under cross-validation.** Its CV F1 and test F1 differ by a small margin, ruling
   out overfitting.
4. **Tunable to the business need.** A single threshold parameter moves it along the
   precision–recall curve without retraining.

**Why not Logistic Regression?** It achieves the highest recall of any model, but at markedly
lower precision — a large share of the individuals it flags as high earners are not. F1 balances
the two, and on that balance XGBoost wins.

**The trade-off, stated plainly.** XGBoost costs interpretability. Logistic Regression hands you
one coefficient per feature; XGBoost hands you hundreds of trees and a gain score. Where an
organisation must justify individual decisions to the people affected by them, that difference
may well outweigh the performance gap.

## 10.3 Business implications

- **Education is the actionable lever.** `education.num` is the strongest numeric predictor and
  one of the very few factors an individual or a programme can genuinely change. Marital status
  predicts well but cannot be the target of an intervention.
- **The model should rank, not decide.** It separates the classes reliably enough to prioritise
  outreach across a large population, but ranking is not a decision. It should route human
  attention, never allocate outcomes on its own.
- **The threshold is a policy choice, not a modelling one.** Lowering it trades precision for
  reach. Where the cost of missing someone exceeds the cost of contacting them unnecessarily,
  the lower threshold is correct — and that is a judgement for the organisation, not the analyst.
- **Occupation clusters justify sector targeting.** Exec-managerial and Prof-specialty roles
  show the highest share of high earners, giving programmes concrete pathways to build toward.

## 10.4 Limitations

1. **The data is thirty years old.** This is a 1994 US Census extract. Occupation categories,
   wage levels and the \$50K threshold are all fixed to that year and do not transfer to the
   present day or to other labour markets.
2. **Protected attributes are used as predictors.** `race`, `sex` and `native.country` all enter
   the model, and `sex` carries a substantial effect size. Any deployment that allocated support
   on the basis of these predictions would require a formal fairness audit first. This analysis
   does not provide one, and that is a genuine limitation rather than a future nicety.
3. **The predictive framing is imperfect.** The stated goal is targeted career support, but the
   model optimises for identifying people who *already* earn well. Its most defensible use is
   factor identification rather than individual targeting.
4. **Importance is not causation.** The dominance of household-structure variables reflects what
   correlates with income in 1994 census data, not what causes it.
5. **Missingness was simplified.** Encoding `?` as a single `Unknown` category preserves the
   signal but flattens whatever distinct reasons produced it.

## 10.5 Future improvements

1. **Fairness analysis** — measure equalised odds and demographic parity across `race` and `sex`,
   then retrain without protected attributes to quantify the performance cost of removing them.
2. **Feature engineering** — construct interaction terms such as education × occupation, and
   binned hours-per-week bands, neither of which the current features capture.
3. **Alternative gradient-boosting libraries** — LightGBM and CatBoost handle categorical
   features natively without one-hot expansion, which would shrink the feature matrix considerably.
4. **Bayesian hyperparameter optimisation** — would cover a wider search space than grid search
   at lower computational cost.
5. **Validation on recent data** — apply the same pipeline to a current census release to test
   whether the relationships identified here still hold three decades on.
